# Inflation trends en Latin America (2000-2025)
## Proyect Objetive: Analyze inflation across select Latin America countries between 2000-2025.
### Note: Argentina and Venezuela were excluded from first part because World Bank Data regarding the indicador for these countries is limited

## Business questions:
1. Which country experienced the hights average inflation?
2. Which country maintained the greatest price stability?
3. Which country exhibited the hights inflation volaty?
4. How did inflation evolve across the region?
5. How does Latin America compare with the United States ?

```python
# Cargar datos
inflation_clean=pd.read_csv("../data/processed/inflation_latam_clean.cvs")

inflation_clean.head(10)
```

```python
#Validaciones iniciales

inflation_clean.shape
```

```python
inflation_clean.info()
```

```python
inflation_clean.describe()
```

```python
#question 1
avg_inflation=(
    inflation_clean.groupby("country")["inflation"].mean().sort_values(ascending=False)
)

avg_inflation
```

## Finding 1
Countries are ranked according to their average inflation rate during the analysis period. Ecuador is the country with the highest average inflation during the 2000–2025 period (8.13%), Uruguay followed closely with an average inflation rate of 8.00%. whereas the United States was the country with the lowest average inflation (2.57%). 

Overall, the results indicate that Latin American economies experienced persistently higher inflation rates than the United States during the period analyzed.


```python
#question 2
inflation_stability=(
    inflation_clean
    .groupby("country")["inflation"]
    .std()
    .sort_values()
)
inflation_stability
```

#Finding 2
Price stability was evaluated using the standard deviation of annual inflation rates. Although Mexico shows greater stability based on the standard deviation results, it is important to note that its average inflation rate during the study period was the second highest. These findings indicate that Mexico's inflation neither improved nor worsened during the period, remaining at an annual average of 4.72%; it is therefore advisable to evaluate inflation on a year-by-year basis to obtain more precise data regarding the policies implemented.

```python
#question 3. País más volatil 
inflation_volatility=(
    inflation_clean
    .groupby("country")["inflation"]
    .std()
    .sort_values(ascending=False)
)

inflation_volatility
```

```python
inflation_clean[
    inflation_clean["country"]=="ECUADOR"
].sort_values("inflation", ascending=False)
```

# Finding 3
Inflation volatility was measured using the standard deviation of annual inflation rates. Ecuador was the most volatile and also showed the highest average inflation during the period. This result was strongly influenced by the exceptionally high inflation rate observed in 2000 (96.09%), as well as episodes of negative inflation recorded in later years.

 as might have been expected from the previous result, the least volatile country was Mexico.

These observations reflect genuine macroeconomic conditions and were therefore retained in the dataset.

# Finding 4 
Post-Pandemic Inflation Surge
 
Regional inflation accelerated significantly between 2021 and 2022, increasing from 4.85% to 8.63%.
 
Although inflation moderated in 2023, average price growth remained above pre-pandemic levels, suggesting persistent inflationary pressures across the region.

Notably, the United States also experienced elevated inflation (8.00%), suggesting that the inflationary episode was driven by global factors rather than country-specific conditions alone.
 
These results indicate that inflationary pressures following the COVID-19 pandemic affected both Latin American economies and advanced economies alike.

```python
#Tabla de inflacion de Latam
latam = (
inflation_clean[
inflation_clean["country"] != "United States"
]
.groupby("year")["inflation"]
.mean()
.reset_index()
)

latam.rename(
columns={"inflation": "latam_inflation"},
inplace=True
)
```

```python
#Tabla de EEUU

usa = (
inflation_clean[
inflation_clean["country"] == "United States"
][["year", "inflation"]]
)

usa.rename(
columns={"inflation": "usa_inflation"},
inplace=True
)
```

```python
comparison = latam.merge(
usa,
on="year"
)

comparison.head()
```

```python
#DIFERENCIA DE LA INFLACION DE LATAM Y EEUU
comparison["gap"] = (
comparison["latam_inflation"]
- comparison["usa_inflation"]
)

comparison.head()
```
